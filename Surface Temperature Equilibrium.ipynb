```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# surface-equilibrium temperature ratio
f = np.linspace(0, 1, 1000)
eps = np.linspace(0, 1, 1000)
F, EPS = np.meshgrid(f, eps)
ratio = ((2-F)/(2-EPS))**0.25

plt.style.use('dark_background')
plt.figure(figsize=(6,6))

levels = np.arange(0.84, 1.2, 0.02)
points = [(0.09,0.91), (0.15,0.85)] + [(i,1-i) for i in np.linspace(0.2, 0.95, 14)] + [(0.95,0.05)]

plt.contourf(EPS, F, ratio, levels=200, cmap='bone')
lines = plt.contour(EPS, F, ratio, levels=levels, colors='white')
plt.clabel(lines, manual=points, inline=True, fontsize=12, fmt='%.2f')

plt.scatter(0.95, 0.29, s=50, color='blue')
plt.annotate('Earth', (0.9, 0.32), color='blue')

plt.title(r'$\frac{T_s}{T_e} = \left( \frac{2 - f}{2 - \varepsilon} \right)^{1/4}$', pad=15, fontsize=20)
plt.ylabel(r'Atmospheric absorptivity after reflection, $f$', fontsize=12)
plt.xlabel(r'Atmospheric emissivity, $\varepsilon$', fontsize=12)
plt.tight_layout()
# plt.savefig('surface_equilibrium temperature ratio.png', dpi=120)
```
