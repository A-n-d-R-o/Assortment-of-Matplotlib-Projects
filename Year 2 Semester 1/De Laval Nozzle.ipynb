```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# De Laval nozzle mass flowrate and area ratio

# parameters in SI units
A     = 1    # cross-sectional area
p_0   = 1e5  # stagnation pressure (set to atmospheric pressure)
gamma = 1.4  # heat capactity ratio (for air)
R     = 287  # specific gas constant (dry air)
T_0   = 293  # stagnation temperature (set to 20°C)

# plot initialisation
plt.style.use("dark_background")
fig, ax = plt.subplots(1, 2, figsize=(15,5))
fig.suptitle("De Laval Nozzle", fontsize=30, y=1.1)

# mass flowrate
Ma1 = np.linspace(0, 5, 200)
m_dot = A * p_0 * Ma1 * np.sqrt(gamma / (R * T_0)) * (
    1 + (gamma - 1) / 2 * Ma1**2
)**(
    (1 + gamma) / (2 * (1 - gamma))
)

# ax[0] formatting
ax[0].plot(Ma1, m_dot, color="lightblue", linewidth=3)
ax[0].fill_between(Ma1, m_dot, zorder=10, color="lightblue", alpha=0.5)
ax[0].plot([1, 1], [0, max(m_dot)], linestyle="--", color="yellow")
ax[0].set_xlim(0, 5)
ax[0].set_ylim(0, 250)
ax[0].grid(alpha=0.2)
ax[0].set_title(
    r"Mass Flowrate $\dot{m} = Ap_0 \text{Ma} \sqrt{\dfrac{\gamma}{RT_0}}"
    r"\left(1 + \dfrac{\gamma - 1}{2} \text{Ma}^2 \right)^\dfrac{1 + \gamma}"
    r"{2(1 - \gamma)}$", fontsize=15)
ax[0].set_ylabel(r"Mass flowrate $\dot{m}$ (kg s$^{-1}$)", fontsize=15)
ax[0].set_xlabel(r"Mach number $\text{Ma}$", fontsize=15)

# area ratio
Ma2 = np.linspace(1e-10, 5, 200)
A_ratio = 1 / Ma2 * (
    (1 + (gamma - 1) / 2 * Ma2**2) / ((gamma + 1) / 2)
)**(
    (gamma + 1) / (2 * (gamma - 1))
)

# ax[1] formatting
ax[1].plot(Ma2, A_ratio, color="lightblue", linewidth=3)
ax[1].fill_between(Ma2, A_ratio, zorder=10, color="lightblue", alpha=0.5)
ax[1].plot([0, 1, 1], [1, 1, 0], linestyle="--", color="yellow")
ax[1].set_xlim(0, 3)
ax[1].set_ylim(0, 6)
ax[1].grid(alpha=0.2)
ax[1].set_title(
    r"Area Ratio $\dfrac{A}{A^*} = \dfrac{1}{\text{Ma}} \left[ \dfrac{1 + "
    r"\frac{\gamma - 1}{2} \text{Ma}^2}{(\gamma + 1) / 2} \right]^\dfrac{\gamma"
    r"+ 1}{2(\gamma - 2)}$", fontsize=15
)
ax[1].set_ylabel(r"Cross-sectional area ratio $A / A^*$", fontsize=15)
ax[1].set_xlabel(r"Mach number $\text{Ma}$", fontsize=15)

# plt.savefig("De-Laval.png", bbox_inches="tight", dpi=300)
```
