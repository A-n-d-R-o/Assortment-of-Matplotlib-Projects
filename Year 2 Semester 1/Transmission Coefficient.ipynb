```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# transmission coefficient

# parameters
m = 9.11e-31     # mass (kg)
hbar = 1.055e-34 # reduced Planck constant (J s)
L = 0.75e-9      # barrier width (m)
eV = 1.6e-19     # electron volt (J)
V0 = 6.8 * eV    # barrier height (eV)

# particle energy list
E = np.linspace(1e-30, V0, 500, endpoint=True)
E_ratio = E / V0

with np.errstate(invalid="ignore", divide="ignore"):
    
    # wavenumber list
    k = np.sqrt(2 * m * (V0 - E)) / hbar
    
    # transmission coefficient
    T = 1 / (1 + V0**2 / (4 * E * (V0 - E)) * np.sinh(k * L)**2)

# plot initialisation
plt.style.use("dark_background")
fig, ax = plt.subplots(figsize=(8,6))
ax.set_title("Quantum Tunnelling Transmission Coefficients", fontsize=18)
ax.set_ylabel(r"Transmission coefficient $T$", fontsize=15)
ax.set_xlabel(r"Energy ratio $E / V_0$", fontsize=15)
ax.set_yscale("log")
ax.set_ylim(1e-10, 1)
ax.plot(
    E_ratio, T, color="lightblue", linewidth=3, label=r"$T = \left( 1 + "
    r"\frac{1}{4} \frac{V_0^2}{E (V_0 - E)} \sinh^2 (k_2L)\right)^{-1}$"
) 
ax.fill_between(E_ratio, T, zorder=10, color="lightblue", alpha=0.5)
ax.grid(alpha=0.2)
ax.legend(loc="upper left", fontsize=15)

# plt.savefig("transmission-coefficient.png", bbox_inches="tight", dpi=300)
```
