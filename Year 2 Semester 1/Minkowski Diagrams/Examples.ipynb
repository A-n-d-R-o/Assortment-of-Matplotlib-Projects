```python
mink = Minkowski(lightcone=True, hyperbolas=True, S0=True, show_data=True)
mink.worldline("S1", 0.5, "red", False, True)
mink.worldline("S2", -0.5, "blue", False, True)
mink.animate(["S1", "S2", "S0"], 90, save=True)
```

```python
num_frames = 100
colours = plt.cm.hsv(np.linspace(0, 1, num_frames))
mink = Minkowski(S0=False, show_data=False)
for i in range(num_frames):
    mink.worldline(f"S{i}", i / num_frames, colours[i], grid=True, label=False)
mink.animate(["S99", "S0"], 80, save=True, title="many-worldlines")
```
