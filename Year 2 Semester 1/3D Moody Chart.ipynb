```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# 3D Moody chart

# plot initialisation
plt.style.use("dark_background")
fig = plt.figure(figsize=(9,8))

# title with space for legend
fig.suptitle("3D Moody Chart         ", fontsize=30, y=0.85)

ax = fig.add_subplot(111, projection="3d", computed_zorder=False)
ax.set_xlim(np.log10(500), np.log10(1e8))
ax.set_ylim(np.log10(1e-6), np.log10(0.08))
ax.set_zlim(np.log10(0.001), np.log10(0.05))

for axis in [ax.xaxis, ax.yaxis, ax.zaxis]:
    axis.set_pane_color((0.2, 0.2, 0.25, 1))

ax.view_init(elev=15, azim=-80)

# xy data
Re_list = np.logspace(np.log10(500), np.log10(1e8), 500)
kD_list = np.logspace(np.log10(1e-6), np.log10(0.08), 500)
Re, kD = np.meshgrid(Re_list, kD_list)

# friction coefficients
Cf_lam = 16 / Re
Cf_lam = np.ma.masked_where(Re > 4000, Cf_lam)
Cf_tur = 1.38e-3 * ( 1 + (2e4 * kD + 1e6 / Re)**(1/3) )
Cf_tur = np.ma.masked_where(Re < 3000, Cf_tur)

# transform co-ords to log10 space
X = np.log10(Re)
Y = np.log10(kD)
Z_lam = np.log10(Cf_lam)
Z_tur = np.log10(Cf_tur)

# plot surfaces
ax.plot_surface(
    X, Y, Z_lam, rstride=2, cstride=2, cmap="cool_r", antialiased=False
)
ax.scatter(0, 0, 0, marker="s", s=200, color="aqua", label="Laminar")

ax.plot_surface(
    X, Y, Z_tur, rstride=2, cstride=2, cmap="hot", antialiased=False
)
ax.scatter(0, 0, 0, marker="s", s=200, color="orangered", label="Turbulent")

# Re axis
xticks = [3, 4, 5, 6, 7, 8]
ax.set_xticks(xticks)
ax.set_xticklabels([r"$10^%d$" % i for i in xticks], va="baseline")
ax.set_xlabel(r"Reynolds number $Re$", fontsize=15)

# k/D axis
yticks = [-5, -4, -3, -2]
ax.set_yticks(yticks)
ax.set_yticklabels([r"$10^{%d}$" % i for i in y_ticks], va="baseline")
ax.set_ylabel(r"Relative roughness $k/D$", fontsize=12)

# C_f axis
zticks = [-3, -2.8,-2.6, -2.4, -2.2, -2, -1.8, -1.6, -1.4]
ax.set_zticks(zticks)
ax.set_zticklabels([f"{10**tick:.1g}" for tick in zticks], va="baseline")
ax.set_zlabel(
    "\n\nFriction coefficient " +  r"$C_f$", 
    fontsize=15, labelpad=0.5
)

plt.legend(loc="upper right", fontsize=15)

plt.savefig("3D-Moody-chart", dpi=300, bbox_inches="tight")
```
