```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation, PillowWriter
from IPython.display import HTML
```

```python
# wave equation standard solution
def fc(num, plus=False):
    if num == 1 and plus:
        return '+'
    if num == 1 and not plus:
        return ''
    if num == -1:
        return '-'
    return f'{round(num, 3)}'

samples = 20
y_m = 4
k = 0.3
sign = -1
omega = 1

phis = np.linspace(0, 2*np.pi, samples, endpoint=False)
x = np.linspace(0, 10, 200)
t = np.linspace(0, 10, 200)
x, t = np.meshgrid(x, t)

plt.style.use('dark_background')
fig = plt.figure(figsize=(6,6))
ax = fig.add_subplot(111, projection='3d')

z_offset = -y_m * 1.5
ax.set_zlim(z_offset, y_m)

ax.xaxis.pane.set_visible(False)
ax.yaxis.pane.set_visible(False)
ax.zaxis.pane.set_visible(False)

title = rf'$ y(x,t) = {fc(y_m)} \sin ({fc(k)}x {fc(sign*omega, True)}t + \phi(t))$'
ax.set_title(title, fontsize=15)
ax.set_xlabel('$x$', fontsize=12)
ax.set_ylabel('$t$', fontsize=12)
ax.set_zlabel('$y(x,t)$', fontsize=12)

y_0 = y_m * np.sin(k * x + sign * omega * t + phis[0])

surface = ax.plot_surface(x, t, y_0, cmap=plt.cm.plasma)
contour = ax.contourf(x, t, y_0, zdir='z', offset=z_offset, cmap=plt.cm.plasma)

def update(frame):
    
    global surface, contour
    
    surface.remove()
    contour.remove()
    
    y = y_m * np.sin(k * x + sign * omega * t + phis[frame])
    surface = ax.plot_surface(x, t, y, cmap=plt.cm.plasma)
    contour = ax.contourf(x, t, y, zdir='z', offset=z_offset, cmap=plt.cm.plasma)
    
    return surface,

anim = FuncAnimation(fig, update, frames=samples, blit=False)
# anim.save(f'wave_equation_solution.gif', writer=PillowWriter(fps=20))
plt.close(fig)
HTML(anim.to_jshtml())
```
