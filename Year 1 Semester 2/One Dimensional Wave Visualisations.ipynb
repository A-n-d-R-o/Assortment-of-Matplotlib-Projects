```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation, PillowWriter
from IPython.display import HTML
```

```python
# travelling wave animation
x_range = 6 * np.pi

plt.style.use('dark_background')
fig, ax = plt.subplots(figsize=[10,3])
ax.set_xlim(0, x_range)
ax.set_ylim(-1.2, 1.2)
plt.subplots_adjust(left=0.015, right=0.985, top=0.95, bottom=0.05)
plt.axvline(0)
plt.axvline(x_range)
plt.axhline(-1.2)
plt.axhline(1.2)
ax.axis('off')

t = np.linspace(0, x_range, 1000)
phases = np.linspace(0, x_range, 200, endpoint=False)

wave, = ax.plot([], [], color='cyan', zorder=-1)

offsets = 2 * np.pi * np.arange(-2, 3)
lines = [
    ax.plot([], [], '--', color='white', alpha=0.5, zorder=-2)[0]
    for _ in offsets
]

def update(frame):
    
    phase = phases[frame]
    y = np.sin(t - phase)

    wave.set_data(t, y)

    for line, offset in zip(lines, offsets):
        x = phase + offset
        line.set_data([x,x], [-1.2,1.2])

    return [wave, *lines] # , *dots]

anim = FuncAnimation(fig, update, frames=200, blit=True)

# anim.save('travelling_wave.gif', writer=PillowWriter(fps=20))

plt.close(fig)

HTML(anim.to_jshtml())
```

```python
# wave train animation
x_range = 6 * np.pi

plt.style.use('dark_background')
fig, ax = plt.subplots(figsize=[10,3])
ax.set_xlim(0, x_range)
ax.set_ylim(-1.2, 1.2)
plt.subplots_adjust(left=0.015, right=0.985, top=0.95, bottom=0.05)
plt.axvline(0)
plt.axvline(x_range)
plt.axhline(-1.2)
plt.axhline(1.2)
ax.axis('off')

t = np.linspace(0, x_range, 30)
phases = np.linspace(0, x_range, 200, endpoint=False)

wave = ax.scatter(t, np.zeros_like(t), color='cyan', s=20, zorder=-1)

offsets = 2 * np.pi * np.arange(-2, 3)
lines = [
    ax.plot([], [], '--', color='white', alpha=0.5, zorder=-2)[0]
    for _ in offsets
]

def update(frame):
    
    phase = phases[frame]
    y = np.sin(t - phase)

    wave.set_offsets(np.column_stack((t, y)))

    for line, offset in zip(lines, offsets):
        x = phase + offset
        line.set_data([x,x], [-1.2,1.2])

    return [wave, *lines]

anim = FuncAnimation(fig, update, frames=200, blit=True)

# anim.save('wave_train_1.gif', writer=PillowWriter(fps=20))

plt.close(fig)

HTML(anim.to_jshtml())
```

```python
# another wave train animation
x_range = 6 * np.pi

plt.style.use('dark_background')
fig, ax = plt.subplots(figsize=[10,3])
ax.set_xlim(0, x_range)
ax.set_ylim(-1.2, 1.2)
plt.subplots_adjust(left=0.015, right=0.985, top=0.95, bottom=0.05)
plt.axvline(0)
plt.axvline(x_range)
plt.axhline(-1.2)
plt.axhline(1.2)
ax.axis('off')

t = np.linspace(0, x_range, 20)
phases = np.linspace(0, x_range, 200, endpoint=False)

wave = ax.scatter(t, np.zeros_like(t), color='cyan', s=50, zorder=-1)
for pos in t:
    ax.axvline(pos, linestyle='--', color='white', alpha=0.5, zorder=-2)

def update(frame):
    
    phase = phases[frame]
    y = np.sin(t - phase)

    wave.set_offsets(np.column_stack((t, y)))

    return wave,

anim = FuncAnimation(fig, update, frames=200, blit=True)

# anim.save('wave_train_2.gif', writer=PillowWriter(fps=20))

plt.close(fig)

HTML(anim.to_jshtml())
```

```python
# arrow wave animation
x_range = 6 * np.pi

plt.style.use('dark_background')
fig, ax = plt.subplots(figsize=[10,3])
ax.set_xlim(0, x_range)
ax.set_ylim(-1.2, 1.2)
plt.subplots_adjust(left=0.015, right=0.985, top=0.95, bottom=0.05)
plt.axvline(0)
plt.axvline(x_range)
plt.axhline(-1.2)
plt.axhline(1.2)
ax.axis('off')

t = np.linspace(0, x_range, 50)
phases = np.linspace(0, x_range, 200, endpoint=False)

zero_list = np.zeros_like(t)
x = t
y = zero_list
u = zero_list
v = zero_list

quiver = ax.quiver(
    x, y, u, v,
    color='cyan',
    angles='xy',
    scale_units='xy',
    scale=1,
    width=0.004,
    zorder=-1
)

def update(frame):
    
    phase = phases[frame]
    v = np.sin(t - phase)

    quiver.set_UVC(u, v)
    return quiver,

anim = FuncAnimation(fig, update, frames=200, blit=True)

# anim.save('arrow_wave.gif', writer=PillowWriter(fps=20))

plt.close(fig)

HTML(anim.to_jshtml())
```
