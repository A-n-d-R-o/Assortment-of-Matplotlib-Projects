```python
import random
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.ticker as ticker
from matplotlib.animation import FuncAnimation, PillowWriter
from IPython.display import HTML
```

```python
# Monte Carlo simulation for pi
def monte_carlo(samples=100, interval=1, scale='linear', fps=10, save=False, filetype='gif'):

    plt.style.use('dark_background')
    fig, ax = plt.subplots(ncols=2, figsize=(10, 5), dpi=200)
    fig.suptitle(r'Monte Carlo Estimation for $\pi$', fontsize=22)
    plt.subplots_adjust(left=0.05, right=0.95, top=0.85, bottom=0.1, hspace=0.2, wspace=0.2)
    ax[0].set_aspect('equal')

    # circle plot
    theta = np.linspace(0, 2*np.pi, 250)
    ax[0].plot(np.cos(theta), np.sin(theta), color='white', linewidth=1.5)
    
    ax[0].set(xlim=[-1,1], ylim=[-1,1])
    ax[0].grid(True, which='major', linestyle='-', linewidth=1.2, alpha=0.8)
    ax[0].xaxis.set_major_locator(ticker.MultipleLocator(0.5))
    ax[0].yaxis.set_major_locator(ticker.MultipleLocator(0.5))
    ax[0].grid(True, which='minor', linestyle='--', linewidth=0.8, alpha=0.5)
    ax[0].xaxis.set_minor_locator(ticker.MultipleLocator(0.25))
    ax[0].yaxis.set_minor_locator(ticker.MultipleLocator(0.25))

    circle_title = ax[0].set_title(f'Random Samples', fontsize=18, x=1, ha='right')
    ax[0].set_xlabel('x', fontsize=12)
    ax[0].set_ylabel('y', fontsize=12)

    # graph plot
    ax[1].axhline(np.pi, color='white', linewidth=1.5)
    ax[1].set_xscale(scale)
    ax[1].grid(True, which='major', linestyle='-', linewidth=1.2, alpha=0.5)

    ax[1].set_xlabel('Number of samples', fontsize=12)
    ax[1].set_ylabel(r'$\pi$ estimate', fontsize=12)
    graph_title = ax[1].set_title(rf'$\pi \approx$', fontsize=18, x=0, ha='left')

    ax_pi = ax[1].twinx()
    ax_pi.set_ylim(ax[1].get_ylim())
    ax_pi.set_yticks([np.pi])
    ax_pi.set_yticklabels([r'$\pi$'])
    ax_pi.tick_params(axis='y', length=0)

    # animate Monte Carlo sampling
    n_in, n_out = 0, 0
    pi_list = []
    line, = ax[1].plot([], [], color='red', linewidth=1.5)

    frames = int(samples/interval)

    def init():
        line.set_data([], [])
        return line,
    
    def update(frame):

        if frame == 0 and frames != 1:
            return line,

        nonlocal n_in, n_out

        for _ in range(interval):
            x, y = random.uniform(-1,1), random.uniform(-1,1)
    
            if x**2 + y**2 <= 1:
                n_in += 1
                ax[0].scatter(x, y, s=20, color='blue', zorder=5)
            else:
                n_out += 1
                ax[0].scatter(x, y, s=20, color='red', zorder=5)
                
            n_samples = n_in + n_out
            pi_estimate = 4*n_in/n_samples
            pi_list.append(pi_estimate)

        line.set_data(range(1, n_samples+1), pi_list)
        ax[1].relim()
        ax[1].autoscale_view()
        ax_pi.set_ylim(ax[1].get_ylim())

        circle_title.set_text(f'{n_samples} Random Samples')
        graph_title.set_text(rf'$\pi \approx {pi_estimate:.5g}$')

        return line,

    if filetype == 'png':
        if save:
            fig.savefig(f'Monte_Carlo_{samples}_samples.png', dpi=300, bbox_inches='tight')

        return plt.show()

    elif filetype == 'gif':
        anim = FuncAnimation(fig, update, frames=frames+1, init_func=init, blit=False)
        if save:
            anim.save(f'Monte_Carlo_{samples}_samples.gif', writer=PillowWriter(fps=fps))
        plt.close(fig)
    
        return HTML(anim.to_jshtml())
```

```python
monte_carlo(samples=100, interval=5, scale='log', save=False)
```
