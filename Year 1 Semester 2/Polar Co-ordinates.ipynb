```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
from skspatial.objects import Cylinder, Sphere
```

```python
# co-ordinate system converter functions
def to_polar(coords, system=None):

    x, y = coords[:2]
    
    rho = np.sqrt(x**2 + y**2)
    phi = np.atan2(y, x) % (2 * np.pi)

    if len(coords) == 2:
        return (rho, phi)

    z = coords[2]

    if system == 'cyl':
        return (rho, phi, z)
        
    elif system == 'sph':
        r = np.sqrt(rho**2 + z**2)
        theta = np.atan2(rho, z)
        return (r, theta, phi)

def to_cart(coords, system=None):

    if len(coords) == 2:
        rho, phi = coords
        x = rho * np.cos(phi)
        y = rho * np.sin(phi)
        return (x, y)

    elif system == 'cyl':
        rho, phi, z = coords
        x = rho * np.cos(phi)
        y = rho * np.sin(phi)
        return (x, y, z)

    elif system == 'sph':
        r, theta, phi = coords
        x = r * np.sin(theta) * np.cos(phi)
        y = r * np.sin(theta) * np.sin(phi)
        z = r * np.cos(theta)
        return (x, y, z)
```

```python
# plane polar co-ordinate plotter
def plot_plane_polar(x, y):

    fig = plt.figure(figsize=(6,6))
    ax = fig.add_subplot()

    rho, phi = to_polar((x, y))

    ax.set_xlim(-1.1*rho, 1.1*rho)
    ax.set_ylim(-1.1*rho, 1.1*rho)
    ax.axis('off')

    ax.axhline(0, color='black', linewidth=0.7, alpha=0.8, zorder=-1)
    ax.axvline(0, color='black', linewidth=0.7, alpha=0.8, zorder=-1)

    ax.scatter(x, y, s=20, color='black', zorder=10)

    # Cartesian plots
    ax.quiver(0, 0, 1, 0, color='red', angles='xy', scale_units='xy', scale=1, zorder=1)
    ax.quiver(0, 0, 0, 1, color='limegreen', angles='xy', scale_units='xy', scale=1, zorder=1)
    
    ax.plot([0, x], [0, 0], color='orange', linestyle=':', linewidth=2, alpha=0.5, zorder=-1)
    ax.plot([x, x], [0, y], color='orange', linestyle=':', linewidth=2, alpha=0.5, zorder=-1)

    # polar plots    
    theta = np.linspace(0, 2*np.pi, 300)
    ax.plot(rho*np.cos(theta), rho*np.sin(theta), color='black', linewidth=0.7, alpha=0.5, zorder=-2)

    ax.quiver(x, y, np.cos(phi), np.sin(phi), color='red', angles='xy', scale_units='xy', scale=1, zorder=1)
    ax.quiver(x, y, -np.sin(phi), np.cos(phi), color='limegreen', angles='xy', scale_units='xy', scale=1, zorder=1)

    ax.plot([0, x], [0, y], color='blue', linestyle='--', linewidth=2, alpha=0.5, zorder=0)
    phi_angle = np.linspace(0, phi, 100)
    ax.plot(0.2*rho*np.cos(phi_angle), 0.2*rho*np.sin(phi_angle), color='blue', linestyle='--', linewidth=2, alpha=0.5, zorder=0)

    # plt.tight_layout()
    # plt.savefig('plane_polar.png', dpi=300)

    plt.show();
```

```python
# cylindrical polar co-ordinate plotter
def plot_cylindrical_polar(x, y, z):

    fig = plt.figure(figsize=(10,10))
    ax = fig.add_subplot(111, aspect='equal', projection='3d')

    rho, phi, z = to_polar((x, y, z), system='cyl')
    
    xy_line = [-1.1*rho, 1.1*rho]
    z_line = [-1-0.1*rho, 1.1*z] if z >= 0 else [1.1*z, 1+0.1*rho]
    zeros = [0, 0]
    ax.plot(xy_line, zeros, zeros, color='black', linewidth=0.7, alpha=0.8)
    ax.plot(zeros, xy_line, zeros, color='black', linewidth=0.7, alpha=0.8)
    ax.plot(zeros, zeros, z_line, color='black', linewidth=0.7, alpha=0.8)

    lim = max(xy_line[1], max(z_line))
    ax.set_xlim(-lim, lim)
    ax.set_ylim(-lim, lim)
    ax.set_zlim(-lim, lim)
    ax.axis('off')

    ax.scatter(x, y, z, s=20, color='black', zorder=10)

    # Cartesian plots
    ax.quiver(0, 0, 0, 1, 0, 0, color='red',arrow_length_ratio=0.2, zorder=1)
    ax.quiver(0, 0, 0, 0, 1, 0, color='limegreen', arrow_length_ratio=0.2, zorder=1)
    ax.quiver(0, 0, 0, 0, 0, 1, color='blue', arrow_length_ratio=0.2, zorder=1)
    
    ax.plot([0, x], [0, 0], [0, 0], color='orange', linestyle=':', linewidth=2, alpha=0.5, zorder=-1)
    ax.plot([x, x], [0, y], [0, 0], color='orange', linestyle=':', linewidth=2, alpha=0.5, zorder=-1)
    ax.plot([x,x] ,[y, y], [0, z], color='orange', linestyle=':', linewidth=2, alpha=0.5, zorder=-1)

    # polar plots    
    theta = np.linspace(0, 2*np.pi, 300)
    ax.plot(rho*np.cos(theta), rho*np.sin(theta), 0, color='black', linewidth=0.7, alpha=0.5)
    ax.plot(rho*np.cos(theta), rho*np.sin(theta), z, color='black', linewidth=0.7, alpha=0.5)

    cylinder = Cylinder(point=[0, 0, 0], vector=[0, 0, z], radius=rho)
    cylinder.plot_3d(ax, color='black', alpha=0.05)

    ax.quiver(x, y, z, np.cos(phi), np.sin(phi), 0, color='red',arrow_length_ratio=0.2, zorder=1)
    ax.quiver(x, y, z, -np.sin(phi), np.cos(phi), 0, color='limegreen', arrow_length_ratio=0.2, zorder=1)
    ax.quiver(x, y, z, 0, 0, 1, color='blue', arrow_length_ratio=0.2, zorder=1)

    ax.plot([0, x], [0, y], [0, 0], color='blue', linestyle='--', linewidth=2, alpha=0.5, zorder=0)
    phi_angle = np.linspace(0, phi, 100)
    ax.plot(
        0.2*rho*np.cos(phi_angle), 0.2*rho*np.sin(phi_angle), np.zeros_like(phi_angle), 
        color='blue', linestyle='--', linewidth=2, alpha=0.5, zorder=0
    )
    ax.plot([x, x], [y, y], [0, z], color='blue', linestyle='--', linewidth=2, alpha=0.5, zorder=0)

    # plt.tight_layout()
    # plt.savefig('cylindrical_polar.png', dpi=300)

    plt.show();
```

```python
# spherical polar co-ordinate plotter
def plot_spherical_polar(x, y, z):

    fig = plt.figure(figsize=(10,10))
    ax = fig.add_subplot(111, aspect='equal', projection='3d')

    r, theta, phi = to_polar((x, y, z), system='sph')

    line = [-r, r]
    zeros = [0, 0]
    ax.plot(line, zeros, zeros, color='black', linewidth=0.7, alpha=0.8)
    ax.plot(zeros, line, zeros, color='black', linewidth=0.7, alpha=0.8)
    ax.plot(zeros, zeros, line, color='black', linewidth=0.7, alpha=0.8)

    ax.set_xlim(line)
    ax.set_ylim(line)
    ax.set_zlim(line)
    ax.axis('off')

    ax.scatter(x, y, z, s=20, color='black', zorder=10)

    # Cartesian plots
    ax.quiver(0, 0, 0, 1, 0, 0, color='red',arrow_length_ratio=0.2, zorder=1)
    ax.quiver(0, 0, 0, 0, 1, 0, color='limegreen', arrow_length_ratio=0.2, zorder=1)
    ax.quiver(0, 0, 0, 0, 0, 1, color='blue', arrow_length_ratio=0.2, zorder=1)
    
    ax.plot([0, x], [0, 0], [0, 0], color='orange', linestyle=':', linewidth=2, alpha=0.5, zorder=-1)
    ax.plot([x, x], [0, y], [0, 0], color='orange', linestyle=':', linewidth=2, alpha=0.5, zorder=-1)
    ax.plot([x,x] ,[y, y], [0, z], color='orange', linestyle=':', linewidth=2, alpha=0.5, zorder=-1)

    # polar plots    
    rotation = np.linspace(0, 2*np.pi, 300)
    ax.plot(0, r*np.cos(rotation), r*np.sin(rotation), color='black', linewidth=0.7, alpha=0.5)
    ax.plot(r*np.cos(rotation), 0, r*np.sin(rotation), color='black', linewidth=0.7, alpha=0.5)
    ax.plot(r*np.cos(rotation), r*np.sin(rotation), 0, color='black', linewidth=0.7, alpha=0.5)

    sphere = Sphere([0,0,0], r)
    sphere.plot_3d(ax, color='black', alpha=0.02)

    ax.quiver(
        x, y, z, np.sin(theta)*np.cos(phi), np.sin(theta)*np.sin(phi), np.cos(theta), 
              color='red',arrow_length_ratio=0.2, zorder=1
    )
    ax.quiver(
        x, y, z, np.cos(theta)*np.cos(phi), np.cos(theta)*np.sin(phi), -np.sin(theta), 
        color='limegreen', arrow_length_ratio=0.2, zorder=1
    )
    ax.quiver(
        x, y, z, -np.sin(phi), np.cos(phi), 0, color='blue', arrow_length_ratio=0.2, zorder=1
    )

    ax.plot([0, x], [0, y], [0, z], color='blue', linestyle='--', linewidth=2, alpha=0.5, zorder=0)
    ax.plot([0, x], [0, y], [0, 0], color='black', linestyle=':', linewidth=2, alpha=0.5)

    theta_angle = np.linspace(0, theta, 100)

    theta_x = 0.2 * r * np.sin(theta_angle) * np.cos(phi)
    theta_y = 0.2 * r * np.sin(theta_angle) * np.sin(phi)
    theta_z = 0.2 * r * np.cos(theta_angle)
    
    ax.plot(theta_x, theta_y, theta_z, color='blue', linestyle='--', linewidth=2, alpha=0.5, zorder=0)
    
    phi_angle = np.linspace(0, phi, 100)
    ax.plot(
        0.2*r*np.cos(phi_angle), 0.2*r*np.sin(phi_angle), np.zeros_like(phi_angle), 
        color='blue', linestyle='--', linewidth=2, alpha=0.5, zorder=0
    )

    # plt.tight_layout()
    # plt.savefig('spherical_polar.png', dpi=300)

    plt.show();
```
