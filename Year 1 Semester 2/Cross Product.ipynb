```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d.art3d import Poly3DCollection
from matplotlib.animation import FuncAnimation, PillowWriter
from IPython.display import HTML
```

```python
# cross product animation
frames = 10

fig = plt.figure(figsize=(6,6))
ax = fig.add_subplot(111, projection='3d')

ax.set_xlim([-1.1, 1.1]) # limits of 1.25 better for parallelogram version
ax.set_ylim([-1.1, 1.1])
ax.set_zlim([-1.1, 1.1])
ax.axis('off')

line = [-2, 2]
zeros = [0, 0]
ax.plot(line, zeros, zeros, color='black', linewidth=0.7, alpha=0.8)
ax.plot(zeros, line, zeros, color='black', linewidth=0.7, alpha=0.8)
ax.plot(zeros, zeros, line, color='black', linewidth=0.7, alpha=0.8)

t = np.linspace(0, 2*np.pi, 200)
ax.plot(np.cos(t), np.sin(t), np.zeros_like(t), color='black', linewidth=0.5, alpha=0.5)

a_list = [1, 0, 0]
a = np.asarray(a_list)

v_a = ax.quiver(0, 0, 0, *a_list, color='red', linewidth=2, arrow_length_ratio=0.2)
v_b = ax.quiver(0, 0, 0, *a_list, color='blue', linewidth=2, arrow_length_ratio=0.2)
v_c = ax.quiver(0, 0, 0, 0, 0, 0, color='green', linewidth=2, arrow_length_ratio=0.2)

# verts = [[[0, 0, 0], [1, 0, 0], [2, 0, 0], [1, 0, 0]]]

# parallelogram = Poly3DCollection(verts)
# ax.add_collection3d(parallelogram)

thetas = np.linspace(0, 2*np.pi, frames, endpoint=False)

def update(frame):
    
    global v_b, v_c # , parallelogram

    theta = thetas[frame]

    b = np.array([np.cos(theta), np.sin(theta), 0])
    c = np.cross(a, b)

    v_b.remove()
    v_c.remove()
    # parallelogram.remove()

    v_b = ax.quiver(0, 0, 0, *b, color='blue', linewidth=2, arrow_length_ratio=0.2)
    v_c = ax.quiver(0, 0, 0, *c, color='green', linewidth=2, arrow_length_ratio=0.2)

    # b_list = b.tolist()
    # verts = [[[0, 0, 0], a_list, (a + b).tolist(), b_list]]

    # parallelogram = Poly3DCollection(verts, facecolor='green', alpha=0.2, edgecolor='none')
    # ax.add_collection3d(parallelogram)

    return v_b, v_c # , parallelogram

anim = FuncAnimation(fig, update, frames=frames, blit=True)
# anim.save('cross_product.gif', writer=PillowWriter(fps=20))
# # anim.save('cross_product_parallelogram.gif', writer=PillowWriter(fps=20))
plt.close(fig)
HTML(anim.to_jshtml())
```

```python
# cross product animation (parallelogram)
frames = 10

fig = plt.figure(figsize=(6,6))
ax = fig.add_subplot(111, projection='3d')

ax.set_xlim([-1.25, 1.25])
ax.set_ylim([-1.25, 1.25])
ax.set_zlim([-1.25, 1.25])
ax.axis('off')

line = [-2, 2]
zeros = [0, 0]
ax.plot(line, zeros, zeros, color='black', linewidth=0.7, alpha=0.8)
ax.plot(zeros, line, zeros, color='black', linewidth=0.7, alpha=0.8)
ax.plot(zeros, zeros, line, color='black', linewidth=0.7, alpha=0.8)

t = np.linspace(0, 2*np.pi, 200)
ax.plot(np.cos(t), np.sin(t), np.zeros_like(t), color='black', linewidth=0.5, alpha=0.5)

a_list = [1, 0, 0]
a = np.asarray(a_list)

v_a = ax.quiver(0, 0, 0, *a_list, color='red', linewidth=2, arrow_length_ratio=0.2)
v_b = ax.quiver(0, 0, 0, *a_list, color='blue', linewidth=2, arrow_length_ratio=0.2)
v_c = ax.quiver(0, 0, 0, 0, 0, 0, color='green', linewidth=2, arrow_length_ratio=0.2)

verts = [[[0, 0, 0], [1, 0, 0], [2, 0, 0], [1, 0, 0]]]

parallelogram = Poly3DCollection(verts)
ax.add_collection3d(parallelogram)

thetas = np.linspace(0, 2*np.pi, frames, endpoint=False)

def update(frame):
    
    global v_b, v_c, parallelogram

    theta = thetas[frame]

    b = np.array([np.cos(theta), np.sin(theta), 0])
    c = np.cross(a, b)

    v_b.remove()
    v_c.remove()
    parallelogram.remove()

    v_b = ax.quiver(0, 0, 0, *b, color='blue', linewidth=2, arrow_length_ratio=0.2)
    v_c = ax.quiver(0, 0, 0, *c, color='green', linewidth=2, arrow_length_ratio=0.2)

    b_list = b.tolist()
    verts = [[[0, 0, 0], a_list, (a + b).tolist(), b_list]]

    parallelogram = Poly3DCollection(verts, facecolor='green', alpha=0.2, edgecolor='none')
    ax.add_collection3d(parallelogram)

    return v_b, v_c, parallelogram

anim = FuncAnimation(fig, update, frames=frames, blit=True)
# anim.save('cross_product_parallelogram.gif', writer=PillowWriter(fps=20))
plt.close(fig)
HTML(anim.to_jshtml())
```
