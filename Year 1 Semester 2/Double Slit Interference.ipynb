```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation, PillowWriter
from IPython.display import HTML
```

```python
# double slit animation
samples = 20
pad = 8
cmap = 'plasma'

times = np.linspace(0, 2*np.pi, samples, endpoint=False)
x = np.linspace(0, 30, 50)
y = np.linspace(-15, 15, 50)

X, Y = np.meshgrid(x, y)

plt.style.use('dark_background')
fig = plt.figure(figsize=(6,6))

ax = [[None, None], [None, None]]

ax[0][0] = fig.add_subplot(2, 2, 1)
ax[0][1] = fig.add_subplot(2, 2, 2)

ax[1][0] = fig.add_subplot(2, 2, 3, projection='3d')
ax[1][1] = fig.add_subplot(2, 2, 4, projection='3d')

fig.subplots_adjust(
    left=0.005, right=0.995,
    bottom=0.005, top=0.995,
    wspace=-0.1, hspace=-0.1
)

def update(frame):
    
    time = times[frame]

    Z1 = np.sin(np.sqrt(X**2 + (Y+8)**2) - time)
    Z2 = np.sin(np.sqrt(X**2 + (Y-8)**2) - time)
    Z_overlap = np.where(Z1 > Z2, Z1, Z2)
    Z3 = Z1 + Z2

    ax[0][0].cla()
    ax[0][1].cla()
    ax[1][0].cla()
    ax[1][1].cla()

    # ax[0][0].text(
    #     15, 17, 'Contour Overlap', ha='center', fontsize=12
    # )
    # ax[0][1].text(
    #     15, 17, 'Contour Superpose', ha='center', fontsize=12
    # )
    # ax[1][0].text(
    #     15, 0, 8, 'Surface Overlap', ha='center', fontsize=12
    # )
    # ax[1][1].text(
    #     15, 0, 8, 'Surface Superpose', ha='center', fontsize=12
    # )

    ax[0][0].contourf(X, Y, Z_overlap, levels=50, cmap=cmap)
    ax[0][1].contourf(X, Y, Z3, levels=50, cmap=cmap)

    ax[1][0].plot_surface(X, Y, Z_overlap, cmap=cmap, linewidth=0)
    ax[1][1].plot_surface(X, Y, Z3, cmap=cmap, linewidth=0)

    for a in ax[0]:
        a.set_aspect('equal', adjustable='box')
        a.set_xlim(x.min()-pad, x.max()+pad)
        a.set_ylim(y.min()-pad, y.max()+pad)
        a.set_axis_off()

    for a in ax[1]:
        a.set_zlim(-3, 3)
        a.view_init(elev=60, azim=-70)
        a.set_axis_off()

        for axis in (a.xaxis, a.yaxis, a.zaxis):
            axis.pane.fill = False
            axis.pane.set_edgecolor('none')
        a.grid(False)

    return []

anim = FuncAnimation(fig, update, frames=samples, blit=False)
# anim.save(f'slit_interference.gif', writer=PillowWriter(fps=20))
plt.close(fig)
HTML(anim.to_jshtml())
```
