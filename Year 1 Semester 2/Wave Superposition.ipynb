```python
import numpy as np
from fractions import Fraction
import matplotlib.pyplot as plt
from matplotlib.ticker import MultipleLocator, FuncFormatter
from matplotlib.animation import FuncAnimation, PillowWriter
from IPython.display import HTML
```

```python
# wave superposition animation (simple)
samples = 100

x = np.linspace(0, 6*np.pi, 1000)
times = np.linspace(0, 2*np.pi, samples, endpoint=False)

plt.style.use('default')
fig, ax = plt.subplots(figsize=[10,4])
ax.set_xlim(0, 6*np.pi)
ax.set_ylim(-2.2, 2.2)

ax.grid(True, which='major', linestyle='-', linewidth=0.8)

wave_1, = ax.plot([], [])
wave_2, = ax.plot([], [])
sum_wave, = ax.plot([], [])

def update(frame):

    time = times[frame]

    y_1 = np.sin(x + time)
    y_2 = np.sin(x - time)
    sum_y = y_1 + y_2

    wave_1.set_data(x, y_1)
    wave_2.set_data(x, y_2)
    sum_wave.set_data(x, sum_y)

    return [wave_1, wave_2, sum_wave]

anim = FuncAnimation(fig, update, frames=samples, blit=True)
plt.close(fig)
HTML(anim.to_jshtml())
```

```python
# helper functions
def joint_period(omega1, omega2):
    if omega1 == 0 or omega2 == 0:
        return 2 * np.pi
    ratio = Fraction(omega1 / omega2).limit_denominator(1000)
    joint_period = 2 * np.pi * ratio.numerator / omega1
    return joint_period

def make_title(A, k, omega, phi):
    
    if A == 0:
        return '0'
    elif A == -1:
        fA = '-'
    elif A == 1:
        fA = ''
    else:
        fA = f'{round(A, 3)}'
        
    if k == 0:
        fk = ''
    elif k == -1:
        fk = '-x'
    elif k == 1:
        fk = 'x'
    else:
        fk = f'{round(k, 3)}x'

    if omega == 0:
        fomega = ''
    elif omega == -1:
        fomega = '-t'
    elif omega == 1:
        if fk == '':
            fomega = 't'
        else:
            fomega = '+t'
    else:
        if fk == '':
            fomega = f'{round(omega, 3)}t'
        elif omega < 0:
            fomega = f'{round(omega, 3)}t'
        else:
            fomega = f'+{round(omega, 3)}t'

    if phi == 0:
        fphi = ''
    elif phi < 0:
        fphi = f'{round(phi, 3)}'
    else:
        fphi = f'+{round(phi, 3)}'

    title = f'{fA} \sin ({fk}{fomega}{fphi})'
    
    return title
    
def set_pi_ticks(x, pos):
    n = x / np.pi
    if np.isclose(n, 0):
        return '0'
    elif np.isclose(n, -1):
        return r'$-\pi$'
    elif np.isclose(n, 1):
        return r'$\pi$'
    else:
        return rf'${n:.1g}\pi$'
```

```python
# wave superposition animation (function)
def superpose(
    A1=1, k1=1, omega1=1, phi1=0, A2=1, k2=1, omega2=-1, phi2=0, samples=100, 
    t=None, xlim=3*np.pi, ylim=None, pi_ticks=True, save=False, gif_name=''
):

    if ylim is None:
        ylim = 1.2 * (A1 + A2)

    x = np.linspace(-xlim, xlim, 1000)

    if t is None:
        t = 2 * joint_period(omega1, omega2)
        
    times = np.linspace(0, t, samples, endpoint=False)

    plt.style.use('dark_background')
    fig, ax = plt.subplots(figsize=[10,4])
    ax.set_xlim(-xlim, xlim)
    ax.set_ylim(-ylim, ylim)
    plt.subplots_adjust(left=0.04, right=0.96, top=0.9, bottom=0.09)

    if pi_ticks:
        ax.xaxis.set_major_locator(MultipleLocator(np.pi))
        ax.xaxis.set_minor_locator(MultipleLocator(np.pi / 4))
        ax.xaxis.set_major_formatter(FuncFormatter(set_pi_ticks))
        
        ax.yaxis.set_minor_locator(MultipleLocator(0.5))
        ax.yaxis.set_major_locator(MultipleLocator(1))
    
    ax.grid(True, which='major', linestyle='-', linewidth=0.8, alpha=0.3)
    ax.grid(True, which='minor', linestyle=':', linewidth=0.5, alpha=0.5)
    
    wave_1, = ax.plot([], [], color='darkred')
    wave_2, = ax.plot([], [], color='darkblue')
    sum_wave, = ax.plot([], [], color='lawngreen', linewidth=2)

    title1 = rf'$y_1 = {make_title(A1, k1, omega1, phi1)}$'
    title2 = rf'$y_2 = {make_title(A2, k2, omega2, phi2)}$'
    
    ax.text(0.02, 1.02, title1, transform=ax.transAxes, 
            ha='left', va='bottom', fontsize=15, color='red')
    
    ax.text(0.98, 1.02, title2, transform=ax.transAxes,
            ha='right', va='bottom', fontsize=15, color='blue')

    ax.text(0.5, 1.02, r'$y_\text{sum} = y_1 + y_2$', transform=ax.transAxes,
            ha='center', va='bottom', fontsize=20, color='lawngreen')
    
    def update(frame):
    
        time = times[frame]
    
        y_1 = A1 * np.sin(k1 * x + omega1 * time + phi1)
        y_2 = A2 * np.sin(k2 * x + omega2 * time + phi2)
        sum_y = y_1 + y_2
    
        wave_1.set_data(x, y_1)
        wave_2.set_data(x, y_2)
        sum_wave.set_data(x, sum_y)
    
        return [wave_1, wave_2, sum_wave]
    
    anim = FuncAnimation(fig, update, frames=samples, blit=True)
    if save:
        anim.save(f'{gif_name}.gif', writer=PillowWriter(fps=20))
    plt.close(fig)
    
    return HTML(anim.to_jshtml())
```

```python
superpose(samples=50, t=None, k1=1.2, omega1=-0.2, omega2=0.1, 
          xlim=10*np.pi, pi_ticks=False, save=False, gif_name='beat_superposition')
```
