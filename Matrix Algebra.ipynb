```python
# matrix algebra properties calculator
def order(mat):

    n_rows = len(mat)
    n_cols = len(mat[0])

    for row in mat:
        if len(row) != n_cols:
            raise ValueError('Invalid matrix.')

    return n_rows, n_cols

def mat_add(mat1, mat2):

    n_rows, n_cols = order(mat1)

    if (n_rows, n_cols) != order(mat2):
        raise ValueError('Orders disagree.')
    
    return [
        [mat1[i][j] + mat2[i][j] for j in range(n_cols)]
        for i in range(n_rows)
    ]

def scalar_mult(n, mat):

    n_rows, n_cols = order(mat)
    
    return [
        [n * mat[i][j] for j in range(n_cols)]
        for i in range(n_rows)
    ]

def mat_mult(mat1, mat2):

    n_rows_1, n_cols_1 = order(mat1)
    n_rows_2, n_cols_2 = order(mat2)

    if n_cols_1 != n_rows_2:
        raise ValueError('Inner dimensions disagree.')

    return [
        [sum([mat1[i][s] * mat2[s][j] for s in range(n_cols_1)]) 
        for j in range(n_cols_2)] for i in range(n_rows_1)
    ]
```

```python
# determinant calculator
def sub_mat(M, row, col):

    n_rows = len(M)
    n_cols = len(M[0])
    
    return [
        [M[i][j] for j in range(n_cols) if j != col]
        for i in range(n_rows) if i != row
    ]

def det(M):

    n_rows = len(M)
    n_cols = len(M[0])

    if n_rows == 1 and n_cols == 1:
        return M[0][0]

    for row in range(n_rows):
        if len(M[row]) != n_cols:
            raise ValueError('Not a square matrix.')

    determinant = 0
    for col in range(n_cols):
        determinant += (-1)**col * M[0][col] * det(sub_mat(M, 0, col))

    return determinant
```
